```python
import numpy as np

prices  = np.array([12.5, 8, 45, 23.9, 5.5, 31, 17.25, 60, 9.99, 38])   # euros per book
morning = np.array([3, 1, 0, 2, 5, 1, 2, 0, 4, 1])    # copies sold before noon
evening = np.array([2, 2, 1, 0, 3, 0, 1, 1, 2, 2])    # copies sold after noon

len(prices[prices > 20])
len(prices[(prices <= 30) & (prices >= 10)])
len(prices[(prices < 9) | (prices > 40)])
(prices*(morning+evening)).sum()
(prices*0.8).sum()
```

```python
import numpy as np

np.random.seed(99)
ratings = np.random.randint(1, 11, (6, 5))   # 6 friends (rows) x 5 movies (columns)

friends = ["Ava", "Ben", "Cleo", "Dan", "Eli", "Fay"]
movies  = ["Galaxy Rush", "Pixel Pirates", "Cake Wars", "Midnight Llama", "Turbo Turtles"]

print(ratings)

movies[ratings.mean(axis = 0).argmax()]

friends[ratings.mean(axis = 1).argmin()]

len(ratings[ratings > 8])

ratings[ratings > 8].sum()

np.median(ratings[0])
```
